### Dictionary

Dictionaries are unordered collections of items. They store data in key-value pairs.
Keys must be unique and immutable (e.g., strings, numbers, or tuples), while values can be of any type.

```python
## creating an empty dictionary
my_dict = {}
print(type(my_dict))
```

```python
# ways to create dictionary
Student={
    "Name":"Prasad",
    "age":22,
    "Gender":"Male"
}
print(Student)

dict2=dict()
print(dict2)
```

```python
## acessing items 
print(Student["Name"])
print(Student['age'])

## accessing by get() method
print(Student.get('Gender'))
```

```python
# Removing items
dict1={
    "col1":"yellow",
    "col2":"red",
    "col3":"orange",
    "col4":"blue",
    "col5":"green"
}

dict1.popitem()  # pop last item
dict1

del dict1["col1"] # pop given key value pair
print(dict1)
```

```python
# del key word an also completely delete an dictionary
del dict1
print(dict1) #this will cause an error because "dict1" no longer exists.
```

```python
##methodes in dictionary

keys=Student.keys() ## get all keys
print(keys)

values=Student.values()  ## get all values
print(values)

items=Student.items() ## get all key-value pair
print(items)
```

```python
student1['Name']="PBB"
print(student1)
print(Student)
```

```python
Student2=Student.copy()      ## shallow copy
Student2['age']=25
print(Student2)
print(Student)
```

```python
## iteration over dictionary
# iteration over keys

for keys in Student.keys():
    print(keys)
```

```python
# iteration over values

for value in Student.values():
    print(value)
```

```python
# iteration over key-value pair

for key,value in Student.items():
    print(f"{key}:{value}")
```

```python
## Nested Disctionaries
students={
    "student1":{"name":"Krish","age":32},
    "student2":{"name":"Peter","age":35}
}
print(students)
```

```python
## Access nested dictionaries elementss
print(students["student1"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
## Iterating over nested dictionaries
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary Comphrehension
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## Condition dictionary comprehension
evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## merge 2 dictionary in 1
dic1={"a":1,"b":2}
dic2={"c":3,"d":4}
merged_dict={**dic1,**dic2}  
print(merged_dict)
```
