## List
- Lists are ordered, mutable collections of items.
- They can contain items of different data types.

```python
## creating list
lis=[]
print(type(lis))
```

```python
mix_list=["prasad",3.1,15,'p']
mix_list
```

```python
fruits=["Apple","Banana","Mango","Orange"]
#Accessing list elements
print(fruits[0])
print(fruits[1])
print(fruits[-1])

# By loop
for fruit in fruits:
    print(fruit)
    
## Iterating with index
for index,fruits in enumerate(fruits):
    print(index,fruits)
    
```

```python
#modifing list
fruits[1]="Watermelon"
fruits
```

```python
# add items in list
fruits.append("Banana")           # add element at last
fruits
```

```python
# insert at specific index
pra=[1,2,3,6,4]
pra.insert(1,4)   
pra       # insert 4 at index 1
```

```python
# remove elements
pra.remove(4)  # remove first occurrence of 4
pra
```

```python
fruits.pop()                   # remove last element
fruits
```

```python
rem_fruit=fruits.pop()
print(rem_fruit)               # print removed element
fruits
```

```python
# methods in list
print(fruits.count("Banana"))  # count occurrences of an element
print(fruits.index("Mango"))   # find index of an element

num=[4,3,5,1,7,2,6]
num.sort()                   # sort the list
print(num)
a=len(num)                # length of the list
print(a)

b=max(num)                # maximum element in the list
print(b)
c=min(num)                # minimum element in the list
print(c)

copy=fruits.copy()          # copy the list
print(copy)

reversed_fruits=list(reversed(fruits))  # reverse the list
print(reversed_fruits)

num.clear()                # clear the list
print(num)                 # print empty list
```

```python
#slicing
fruits
print(fruits[1:3])          # slice from index 1 to 2
print(fruits[:3])           # slice from start to index 2   
print(fruits[2:])           # slice from index 2 to end
print(fruits[::2])
print(fruits[::-1])
print(fruits[-3:])
print(fruits[-3:-1])

print(fruits)
```

#### List compresension

```python
num= [x**2 for x in range(10)]
print(num)
```

```python
lst=[]
for x in range(10):
    lst.append(x**2)
print(lst)
```

```python
### List Comprehension with Condition
lst=[]
for i in range(10):
    if i%2==0:
        lst.append(i)

print(lst)
```

```python
## List Comprehension with function calls
words = ["hello", "world", "python", "list", "comprehension"]
lengths = [len(word) for word in words]
print(lengths)  # Output: [5, 5, 6, 4, 13]
```

```python
## Nested List Comphrension

lst1=[1,2,3,4]
lst2=['a','b','c','d']

pair=[[i,j] for i in lst1 for j in lst2]

print(pair)
```
