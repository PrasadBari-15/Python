#### String

```python
##Assign String to variable
name="prasad"
print(name)
```

```python
## multiline String
a="""hello 
            my name is
                            Prasad"""
                            
print(a)
```

```python
## Strings are array/Access String
str1="Hello"
print(str1[1])
```

```python
## loops in string

for i in str1:
    print(i)
```

```python
## String length
print(len(str1))
```

```python
## Check presence of element in string

print("H" in str1)
print('t' in str1)
```

```python
## Slicing in string
print(str1[:3])
print(str1[1:])
print(str1[::-1])
print(str1[::2])
print(str1[-4:-1])
```

```python
## Method in string
#split
a="prasad.bari.15.03.2003"
print(a.split("."))
```

```python
#occurence
a="sea ohh sea what a sea"
print("Occurance of sea:",a.count("sea"))
```

```python
#remove
c="F.R.I.E.N.D"
print(c.replace(".",""))
```

```python
#Ascending order (sort)
a="hello"
b=sorted(a)
print(b)
```

```python
a="hello world"

print(a.find("hello"))   #0
print(a.upper())  
print(a.lower())
print(a.replace("hello","welcome"))
print()
```

```python
## Concatenation
a="hello"
b="prasad"
c=a+b
print(c)
```

```python
# Format String
age=22
txt=f"<y age is {age}"
print(txt)

print(f"Hello im {age} year old")
```
