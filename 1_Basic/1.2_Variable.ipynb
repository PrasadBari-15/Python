## Variables
Variables are fundamental elements in programming used to store data that can be referenced and manipulated in a program. In Python, variables are created when you assign a value to them, and they do not need explicit declaration to reserve memory space. The declaration happens automatically when you assign a value to a variable.

```python
## Declaring and Assigning Variables

age=22
height=5.9
name="Prasad"
is_student=True

# Printing the variables
print("Age:", age)
print("Height:", height)
print("Name:", name)
print("Is Student:", is_student)
```

```python
##Naming Conventions
# 1. Variable names can contain letters, numbers, and underscores.
# 2. Variable names cannot start with a number.
# 3. Variable names are case-sensitive (age and Age are different).
# 4. Avoid using reserved keywords (like `if`, `else`, `while`, etc.) as variable names.
```

```python
# valid names
first_name ="Prasad"
Last_name ="Bari"
age_12=22

##Invalid names
# 1st_name = "Prasad"  # Cannot start with a number
# @Rupees =100 # Cannot start with a special character

#Case Sensitivity
name = "Prasad"
Name = "Bari"
```

### Variable Types
Python dynamically typed , type of a cariable is determined at run time

```python
age=25
name = "Prasad"
height = 5.9
is_student = True

print(type(age))
print(type(name))
print(type(height))
print(type(is_student))
```

```python
## Type Conversion
#1. using 2nd variable
print(age)
print(type(age))
age_str =str(age)
print(age_str)
print(type(age_str))
```

```python
#without using 2nd variable
age = '22'
print(age)
age = int(age)
print(type(age))
print(age,type(float(age)))

name="prasad"
# print(int(name))  # This will raise an error because 'name' is a string and cannot be converted to int
```

```python
## Dynamic Typing
## Python allows the type of a varible to change as the program executes
var=10 #int
print(var,type(var))

var="Hello"
print(var,type(var))

var=3.14
print(var,type(var))

```
