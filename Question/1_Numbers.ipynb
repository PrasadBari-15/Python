##### 1. Find Number is odd or even

```python
def Odd_Even(num):
    if num==0:
       print("number is zero") 
    elif num%2==0:
        print("number is Even")
    else:
        print("Number is odd")
    
    
        
Odd_Even(2)
Odd_Even(0)
```

#### 2. Find Number is positive or negative

```python
def Find(num):
    if num==0:
        print("Number is zero")
    elif num>0:
        print("number is Positive")
    else:
        print("Number is Negative")
        
Find(1)
Find(-6)
Find(0)
```

#### 5. Find number is prime or not

```python
def Prime(num):
    if num==1 or num<=0 :
        print("It is not a prime number")
    else:
        for i in range(2,num):
            if num%i==0:
                print("Number is not Prime")
                break
        else:
            print("Number is Prime")
                
Prime(8)
Prime(7)
Prime(1)
```

#### 6. Fibonacci series

```python
## print series 
def Fib(num):
    a=0
    b=1
    print(a)
    print(b)
    for i in range(2,num):
        c=a+b
        print(c)
        a=b
        b=c
    # print(f"{num} of series is ",c)

Fib(8)
```

```python
## by recursion 
def Fib(num):
    if num<=0:
        return 0
    elif num==1:
        return 1
    else:
        return Fib(num-1)+Fib(num-2)

F= [ Fib(i) for i in range(10) ]
print(F)

```

#### 7. Factorial

```python
# by recursion
def Fact(num):
    if num<=0:
        return 0
    elif num==1:
        return 1
    else:
        return num*Fact(num-1)

Fact(5)
```

#### 9. Reverse a Number

```python
def NUMBER(num):
    rev=0
    while(num>0):
        div=num%10
        rev = rev *10 + div
        num=num//10
    return rev

print(NUMBER(89))
    
        
```

#### 10. Palindrome number

```python
def palindrome(num):
    orig_num = num
    rev_num = 0       

    while num > 0:  
        digit = num % 10   
        rev_num = (rev_num * 10) + digit  
        num=num // 10  
    
    if orig_num==rev_num:
        print(f"{orig_num} is a Palindrome number")
    else:
        print(f"{orig_num} is not Palindrome number")
        
palindrome(55)
palindrome(98)


 
```
