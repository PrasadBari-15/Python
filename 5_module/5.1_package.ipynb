#### Importing Modules in Python: Modules and Packages
In Python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math
math.sqrt(16)
```

```python
from math import sqrt,pi
print(sqrt(16))
print(pi)
```

```python
import numpy as np
np.array([1, 2, 3])
```

```python
## creating own package
'''
    step 1: create a folder eg. package
    step 2: create an file __init.py in package folder
    step 3: create an file where we defining the packages eg. maths,py
'''
from package.maths import add,mul
print(add(4,6))
print(mul(4,5))
```
