1. Instance Method
- Define inside a class , work with an object
- It takes self as first parameter

```python
## methods in class
class Student:
    College="ABC College"              # class Attribute
    def __init__(self,name,*marks):
        self.name=name
        self.marks=marks
    
    def Average(self):
        sum=0
        for i in self.marks:
            sum+=i
        print(f"Hi {self.name}, your average marks are ",sum/len(self.marks))

S3=Student("Prasad",86,97,78)
S3.Average() 
    
```

2. Class Method
- A class method is a method that is bound to the class and not the instance of the class. It can access and modify class-level data (shared by all instances), and is declared using the @classmethod decorator.



```python
class Employee:
    company = "TCS"  # Class variable

    def __init__(self, name):
        self.name = name

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

e1 = Employee("Prasad")
e2 = Employee("Ravi")

print(e1.company)  # TCS
print(e2.company)  # TCS

Employee.change_company("Infosys")  # Using class method to change shared data

print(e1.company)  # Infosys
print(e2.company)  # Infosys
```

```python
#class method
class Student:
    def __init__(self,name,age,division):
        self.name=name
        self.age=age
        self.division=division
    @classmethod
    def from_string(cls, string):
        name, age, division = string.split(',')
        return cls(name, int(age), division)
    def display(self):
        print(f"Name: {self.name}, Age: {self.age}, Division: {self.division}")

# Creating an object using class method
student_str = "John,20,A"
student = Student.from_string(student_str)
student.display()
```

```python
class Student:
    def __init__(self,name,age,division):
        self.name=name
        self.age=age
        self.division=division
    @classmethod
    def from_string(cls, name,age,division):
        cls.name = name
        cls.age = age
        cls.division = division
        return cls(name, age, division)
    def display(self):
        print(f"Name: {self.name}, Age: {self.age}, Division: {self.division}")

# Creating an object using class method

student = Student.from_string("Alice", 21, "B")
student.display()   
print(student.name, student.age, student.division)    
```

3. Static Method 
- Static method is used when the method doesn't use self or cls and is utility-based

```python
class Maths:
    @staticmethod
    def add(x,y):
        return x+y
    
    @staticmethod
    def print():
        print("Hello")
        
a=Maths.add(4,5)
Maths.print()
print(a)
```

```python
# Static Method
class Employee:
    def __init__(self,name,age,role):
        self.name=name
        self.age=age
        self.role=role
        
    @staticmethod
    def hello():
        print("This is an Static method")
    
    def Salary(self,amount=15000):
        self.amount=amount
        if self.role=="Manager":
            self.amount=amount+5000
        elif self.role=="Developer":
            self.amount=amount+3000
        elif self.role=="Tester":
            self.amount=amount+2000
        return f"Salary of {self.name} is {self.amount}"

emp1=Employee("Prasad",22,"Manager")
emp2=Employee("Rohit",25,"Developer")
print(emp1)
print(emp1.name,emp1.age)
print(Employee.hello())
print(emp1.Salary())
print(emp2.Salary())
```

```python
## program that contain all methods

class Employee:
    Company="TCS"                 # Class Attribute
    
    def __init__(self,name,age,role):       # Parameterized constructor
        self.name=name
        self.age=age
        self.role=role
        
    def Salary(self,basic_amount=15000):      ## Instance Class
        self.amount=basic_amount
        if self.role=="Manager":
            self.amount=basic_amount+5000
        elif self.role=="Developer":
            self.amount=basic_amount+3000
        elif self.role=="Tester":
            self.amount==basic_amount+2000
        
    def get_Salary(self):
        return f"Salary of {self.name} is {self.amount}"
    
    @classmethod
    def Change(cls,new_company):
        cls.Company=new_company
    
    @staticmethod
    def Print():
        print("This is an static method")
    

emp1=Employee("Prasad",22,"Manager")
emp1.Salary()
print(emp1.get_Salary())
# Access class attribute correctly
print("Company:", emp1.Company)
Employee.Change("Infosys")  # Changing class attribute using class method
print("Updated Company:", emp1.Company)

Employee.Print() # Calling static method


    
            
```

```python
## Banking System

class Bank:
    Bank="SBI"
    
    def __init__(self,name,Account,Balance):
        self.Customer_name=name
        self.Account_no=Account
        self.Saving_Balance=Balance
    
    def credit(self,amount):
        print("previous balance: ",self.Saving_Balance)
        self.Saving_Balance +=amount
        print(f"Hi {self.Customer_name}, your account has been credited with Rs {amount}.")
        print("Total Balance: Rs ",self.Total())
        
    def debit(self,amount):
        print("previous balance: ",self.Saving_Balance)
        self.Saving_Balance -=amount
        print(f"Hi {self.Customer_name}, your account has been debited with Rs {amount}.")
        print("Total Balance: Rs ",self.Total())
    
    def Total(self):
        return self.Saving_Balance
    

C1=Bank("Prasad",123456,1000)
C1.credit(1000)
C1.credit(50000)

C2=Bank("Kishan",987654,2000)
print(C2.Customer_name)
del C2.Customer_name
#print(C2.Customer_name)   #raise an error
del C2
```
