#### Polymorphism
- poly = many
- morphism = Forms

Sigle function having many formst refers to the use of a single type entity (method, operator or object) to represent different types in different scenarios.

```python
class Cat:
    def __init__(self, name, colour):
        self.name = name
        self.colour = colour

    def info(self):
        print(f"I am a cat. My name is {self.name}. I am {self.colour} years old.")

    def make_sound(self):
        print("Meow")


class Dog:
    def __init__(self, name, colour):
        self.name = name
        self.colour = colour

    def info(self):
        print(f"I am a dog. My name is {self.name}. I am {self.colour} years old.")

    def make_sound(self):
        print("Bark")


cat1 = Cat("Kitty", 2.5)
dog1 = Dog("Fluffy", 4)

cat1.info()
dog1.info()

## Polymorphism in action:
for animal in (cat1, dog1):
    animal.make_sound()
    animal.info()
    animal.make_sound()

```

```python
class Vehicle:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Move!")

class Car(Vehicle):
  pass

class Boat(Vehicle):
  def move(self):
    print("Sail!")

class Plane(Vehicle):
  def move(self):
    print("Fly!")

car1 = Car("Ford", "Mustang") #Create a Car object
boat1 = Boat("Ibiza", "Touring 20") #Create a Boat object
plane1 = Plane("Boeing", "747") #Create a Plane object

for x in (car1, boat1, plane1):
  print(x.brand)
  print(x.model)
  x.move()
```

```python
class vehical:
    def __init__(self,name,brand):
        self.name=name
        self.brand=brand

class car(vehical):
    def move(self):
        print("car is moved")

class bike(vehical):
    def move(self):
        print("bike is running")

V1=bike("Rider","TVS")
V2=car("Defender","Land Rover")

## Polymorphism in action:
for i in(V1,V2):
    i.move()
    print(i.brand)
        
```
