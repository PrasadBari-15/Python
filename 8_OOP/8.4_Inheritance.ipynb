## Inheritance

```python
#Single level Inheritance

class Car:
    @staticmethod
    def start():
        print("Car started")
    @staticmethod
    def stop():
        print("car Stop")
        
class Land_rover(Car):
    @staticmethod
    def model():
        print("Land Rover Model: Defender")
    @staticmethod
    def price():
        print("1 Cr.")
        
# Creating an object of Land_rover class

Car1=Land_rover()
Car1.model()
Car1.price()
Car1.start()
Car1.stop()
```

```python
class Animal:
    def __init__(self, name):
        self.name = name
    
class Dog(Animal):
    def bark(self):
        print(f"{self.name} says Woof!")
        
# Creating an object of Dog class
dog = Dog("Sheru")
dog.bark()
```

```python
class Parent:
    def __init__(self, name):
        self.name = name

    def display(self):
        print(f"Parent: {self.name}")

class Child(Parent):
    def __init__(self, name, age):
        super().__init__(name)  # Call parent's __init__
        self.age = age

    def display(self):
        super().display() # Call parent's display
        print(f"Child: Age: {self.age}")

child = Child("Alice", 10)
child.display()
```

```python
## multilevel Inheritance

class Car:
    @staticmethod
    def start():
        print("Car started")
    @staticmethod
    def stop():
        print("car Stop")
        
class Land_rover(Car):
    @staticmethod
    def model():
        print("Land Rover Model: Defender")
    @staticmethod
    def price():
        print("1 Cr.")

class service(Land_rover):
    @staticmethod
    def service():
        print("Car is modified...")


car2=service() 
car2.service()
car2.model()
car2.start()
```

```python
class car:
    def __init__(self,name,type):
        self.name=name
        self.Type=type

    def Display(self):
        print(f"name of model {self.name}, Type={self.Type}")

class Tata(car):
    def __init__(self,name,type,colour,price):
        self.colour=colour
        self.price=price
        super().__init__(name,type)
    
    def Price(self):
        print(self.price)
        
car1=Tata("Naano","Electric","White","50000")
car1.Display()
car1.Price()
```

```python
class Person:
    def __init__(self,name,age,gender):
        self.name=name
        self.gender=gender
        self.age=age
    
    def Display(self):
        print(self.name)
        print(self.gender)
        print(self.age)
    
class Employee(Person):
    def __init__(self,Id,name,age,gender,role):
        super().__init__(name,age,gender)
        self.Id=Id
        self.role=role
    
    def salary(self,salary=20000):
        if self.role=="Manager":
            self.salary=salary+5000
        elif self.role=="Developer":
            self.salary=salary+3000
        else:
            self.salary=salary
            
        return f"Salary of {self.name} is {self.salary}"

class Student(Employee):
    def __init__(self,College,Id,name,age,gender,role):
        self.college=College
        super().__init__(Id,name,age,gender,role)
    
    def S_Display(self):
        return f"Student name {self.name} work as {self.role} on salary {self.salary} in office studies in {self.college} ."
        

Emp1=Employee(1234,"Prasad",23,"Male","Manager")

print(Emp1.salary())
Emp1.Display()

S1=Student("PRPCEM",1234,"Rohan",22,"Male","Developer")

print(S1.salary())
print(S1.S_Display())
```

```python
## Multiple

class Student:
    @staticmethod
    def Sports():
        print("Sports day is on 15th August")
        
        
class Boy_Games():
    @staticmethod
    def b_game():
        print("Kho-kho","Football","Cricket","Boxing")

class Girl_Games():
    @staticmethod
    def g_game():
        print("Softball","Kho-kho","Dodgeball","Basketball")
    
    
class Boy(Student,Boy_Games):
    @staticmethod
    def play():
        print("All the best for sports day")
        
S1=Boy()
S1.Sports()
S1.b_game()
S1.play()


    
```

```python
class Mother:
    def __init__(self,name):
        self.mothername = name
 
    def mother(self):
        print(f"Mother name {self.mothername}")
        
class Father:
    def __init__(self,name):
        self.fathername=name
    
    def father(self):
        print(f"Father name is {self.fathername}")
        
class Son(Mother,Father):
    def __init__(self,name):
        self.name=name
    
    def son(self):
        print(f"son name {self.name}")
        
class Son2(Mother,Father):
    sonname=""
    
    def son2(self):
        print(f"son name :{self.sonname}")


son1=Son("Ram")
Son.fathername="dashrath"
Son.mothername="Sumitra"

son2=Son2("")

son2.mothername="Kakai"
son2.fathername="Dashrath"
son2.sonname="Bharat"

son1.father()
son1.mother()
son1.son()

son2.father()
son2.mother()
son2.son2()
```

```python
class A:
    @staticmethod
    def a():
        print("from class A")
    
class B(A):
    @staticmethod
    def b():
        print("from class B")

class C(A):
    @staticmethod
    def c():
        print("from class c")
        
class D(B,C):
    @staticmethod
    def d():
        print("from class D")
        

D1=D

D1.d()
D1.c()
D1.a()
```

```python
class c1:
    def hello(self):
        print("hello from c1")
        
class c2(c1):
    def hello(self):
        super().hello()
        print("Hello from class c2")

class c3(c1):
    def hello(self):
        super().hello()
        print("Hello from class c3")
        
class c4(c2,c3):
    def hello(self):
        super().hello()
        print("Hello from class c4")
        
c5=c4()
c5.hello()
```
