### Constructor
- Constructor is a special type of method within a class that is automatically called when a new object of that class is created
- In python, All classes have a function called __init__(), which is always executed when the class is being initiated.

```python
class Student:
    name='prasad'
    ## Default constructor
    def __init__(self):
        print("Constructor called")
        
# Constructor is a special method that is called when an object is created
S2=Student() 
print(S2.name)
```

```python
class Employee:
    #Parameterized constructor
    def __init__(self, name, age, city):
        self.name = name
        self.age = age
        self.city = city

emp1=Employee("John", 30, "pune")
emp2=Employee("Jack", 25, "Mumbai")
emp2.salary=60000

print(emp1.name, emp1.age, emp1.city)
print(emp2.name, emp2.age, emp2.city, emp2.salary)   
```

```python
# Static Method
class Employee:
    def __init__(self,name,age,role):
        self.name=name
        self.age=age
        self.role=role
        
    @staticmethod
    def hello():
        print("This is an Static method")
    
    def Salary(self,amount=15000):
        self.amount=amount
        if self.role=="Manager":
            self.amount=amount+5000
        elif self.role=="Developer":
            self.amount=amount+3000
        elif self.role=="Tester":
            self.amount=amount+2000
        return f"Salary of {self.name} is {self.amount}"

emp1=Employee("Prasad",22,"Manager")
emp2=Employee("Rohit",25,"Developer")
print(emp1)
print(emp1.name,emp1.age)
print(Employee.hello())
print(emp1.Salary())
print(emp2.Salary())
```

```python
class Employee:
    company = "TCS"  # Class variable

    def __init__(self, name):
        self.name = name

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

e1 = Employee("Prasad")
e2 = Employee("Ravi")

print(e1.company)  # TCS
print(e2.company)  # TCS

Employee.change_company("Infosys")  # Using class method to change shared data

print(e1.company)  # Infosys
print(e2.company)  # Infosys
```

```python
#class method
class Student:
    def __init__(self,name,age,division):
        self.name=name
        self.age=age
        self.division=division
    @classmethod
    def from_string(cls, string):
        name, age, division = string.split(',')
        return cls(name, int(age), division)
    def display(self):
        print(f"Name: {self.name}, Age: {self.age}, Division: {self.division}")

# Creating an object using class method
student_str = "John,20,A"
student = Student.from_string(student_str)
student.display()
```

```python
class Student:
    def __init__(self,name,age,division):
        self.name=name
        self.age=age
        self.division=division
    @classmethod
    def from_string(cls, name,age,division):
        cls.name = name
        cls.age = age
        cls.division = division
        return cls(name, age, division)
    def display(self):
        print(f"Name: {self.name}, Age: {self.age}, Division: {self.division}")

# Creating an object using class method

student = Student.from_string("Alice", 21, "B")
student.display()   
print(student.name, student.age, student.division)    
```
