#### Pillers of OOP
 There are 4 main pillers of oop
 1. Abstraction
 2. Encapsulation
 3. Polymorphism
 4. Inheritance

## 1. Abstraction
- Abstraction means hiding the complex implementation details and showing only the essential features.
- Goal: Focus on what an object does rather than how it does it.
- Example: Think of a TV remote—you press a button to increase volume, but don’t know the internal working of the circuit.

### 2. Encapsulation
- Definition: Encapsulation means wrapping data and methods into a single unit (a class) and restricting direct access to some of the object's components.
- Goal: Protect the data from unauthorized access and modification.
- Example: In a bank account, you can deposit or withdraw money, but you can’t directly change the balance.


Program Demonstrating Both

```python
class Bank:
    Bank_Name="SBI"
    def __init__(self,name,amount):
        self.Customer_name=name
        self.saving_amt=amount
    
    def credit(self,amount):
        print("Previous Saving amount is Rs.",self.saving_amt)
        self.saving_amt +=amount
        print(f"Hello {self.Customer_name} amount {amount} is credited Sucessfully")
        print(self.Total())

    def debit(self,amount):
        print("Previous Saving amount is ",self.saving_amt)
        self.saving_amt -=amount
        print(f"Hello {self.Customer_name} amount {amount} is debited Sucessfully")
        print(self.Total())
        
    def Total(self):
        return f"Your current bank balance is Rs. {self.saving_amt}"
    
Cus1=Bank("Prasad",1000)
print(Cus1.Customer_name, Cus1.saving_amt)
    
Cus1.credit(5000)
```
