### Conditional Statement ( if, elif , else )

```python
# if Statement

age = 20

if age>=18:
    print("Able to vote")
```

```python
# else statement

age =16

if age >=18:
    print("Able to vote")

else:
    print("Unable to vote ")
```

```python
# elif Statement

age=int(input("Enter your age"))
print("Entered Age is : ",age)

if age<=13:
    print("You are a child")

elif age <=18:
    print("You are a Teenager")

else :
    print("You are Adult")
    
```

```python
# nested if-else

n=int(input("Enter a number: "))
print("Entered number is :",n)

if n>0:
    print("Number is Positive")
    if n%2==0:
        print("Number is Even")
    else :
        print("Number is odd")

else:
    print("Entered number is Zero or negative")
```

```python
# check year is leap or not

year=int(input("Enter an year:"))
print("Enterd year is :",year)

if year%4==0:
    if year%100==0:
        if year%400==0:
            print("Leap year")
        else:
            print("Non-leap year")
    else:
        print("Leap year")    
else:
    print("Non-leap year")  
```
