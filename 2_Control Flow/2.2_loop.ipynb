### Loops


```python
# For Loop
range(5)
```

```python
for i in range(11):
    print(i)
```

```python
for i in range(1,10,2):
    print(i)
```

```python
# for loop in string
str="Prasad"
for i in str:
    print(i)
```

```python
# nested loops 

for i in range(3):
    for j in range(2):
        print(f"i={i} , j={j}")
```

```python
# while loop

count=1
while count<10:
    print(count)
    
    count=count+1
```

```python
# break 
#The break statement exits the loop

for i in range(10):
    if i==6:
        break
    print(i)
```

```python
##continue

## The continue statement skips the current iteration and continues with the next.

for i in range(10):
    if i==4:
        continue
    print(i)
```

```python
## print sum of first n numbers
n=10
sum=0
count=1
while count<=n:
    sum=sum+count
    count=count+1
print(sum)
```

```python
## print prime numbers in a range
for num in range(100):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)
```
